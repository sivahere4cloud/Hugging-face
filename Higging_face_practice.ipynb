```python
from transformers import AutoTokenizer
MODEL_ID = "distilbert/distilgpt2"
tokenizer =  AutoTokenizer.from_pretrained(MODEL_ID)
print(tokenizer)
```

Practical 2 -Tokenize Text

```python
text = "Siva is learning Hugging Face"
tokens = tokenizer.tokenize(text)
print(tokens)
```

practical - 3  Convert tokens into token ids

```python
token_ids = tokenizer.convert_tokens_to_ids(tokens)
print(token_ids)
```

Practical 6 — Decode

```python
ids = tokenizer.encode(
    "Artificial intelligence is changing software engineering."
)

decoded = tokenizer.decode(ids)

print(ids)
print(decoded)
```

Practical 7 — Inspect individual token ↔ ID mapping

```python
text = "Machine learning"

tokens = tokenizer.tokenize(text)
ids = tokenizer.convert_tokens_to_ids(tokens)

for token, token_id in zip(tokens, ids):
    print(f"{token!r:20} -> {token_id}")
```

```python
text = "Hello world"

without_special = tokenizer.encode(
    text,
    add_special_tokens=False,
)

with_special = tokenizer.encode(
    text,
    add_special_tokens=True,
)

print("Without:", without_special)
print("With:", with_special)
```

```python
tokenizer.pad_token = tokenizer.eos_token
texts = [
    "AI is useful.",
    "Artificial intelligence is changing software engineering rapidly.",
]

encoded = tokenizer(
    texts,
    padding=True,
    return_tensors="pt",
)

print(encoded)

print(encoded["input_ids"].shape)
print(encoded["attention_mask"].shape)
```

```python
#Practical 13 — Inspect tensor shape
text = "I am learning Hugging Face."

encoded = tokenizer(
    text,
    return_tensors="pt",
)

print(encoded["input_ids"])
print(encoded["input_ids"].shape)
```

```python
#Truncation
long_text = "Artificial intelligence " * 1000

encoded = tokenizer(
    long_text,
    truncation=True,
    max_length=50,
)

print(len(encoded["input_ids"]))
```

```python
#Practical 14 — Compare truncation
text = "machine learning " * 100

full = tokenizer(
    text,
    truncation=False,
)

short = tokenizer(
    text,
    truncation=True,
    max_length=20,
)

print("Original length:", len(full["input_ids"]))
print("Truncated length:", len(short["input_ids"]))
```

```python
texts = [
    "Hello",
    "I am learning Hugging Face",
]

dynamic = tokenizer(
    texts,
    padding=True,
)

fixed = tokenizer(
    texts,
    padding="max_length",
    truncation=True,
    max_length=20,
)

print("Dynamic length:", len(dynamic["input_ids"][0]))
print("Fixed length:", len(fixed["input_ids"][0]))
```

```python
examples = [
    "Hello",
    "Hello, how are you today?",
    "Supercalifragilisticexpialidocious",
    "AI 🤖 is amazing!",
]

for text in examples:
    ids = tokenizer.encode(text)

    print("-" * 50)
    print("Text:", text)
    print("Characters:", len(text))
    print("Tokens:", len(ids))
    print("IDs:", ids)
```

```python
print("Tokenizer class:", tokenizer.__class__.__name__)
print("Fast tokenizer:", tokenizer.is_fast)
print("Vocabulary size:", tokenizer.vocab_size)
print("Model max length:", tokenizer.model_max_length)
print("Padding side:", tokenizer.padding_side)
#Right padding
#A B C PAD PAD
#Left padding
#PAD PAD A B C

print("Truncation side:", tokenizer.truncation_side)
```

```python
#Visualize left vs right padding
texts = [
    "Hello",
    "Hello, how are you?",
]

tokenizer.pad_token = tokenizer.eos_token

tokenizer.padding_side = "right"
right = tokenizer(
    texts,
    padding=True,
)

print("RIGHT")
print(right["input_ids"])

tokenizer.padding_side = "left"
left = tokenizer(
    texts,
    padding=True,
)

print("LEFT")
print(left["input_ids"])
```

```python
#inspect whether a tokenizer has a chat template
from transformers import AutoTokenizer

MODEL_ID = "Qwen/Qwen2.5-0.5B-Instruct"

tokenizer = AutoTokenizer.from_pretrained(MODEL_ID)

print(tokenizer.chat_template)
```

```python
messages = [
    {
        "role": "system",
        "content": "You are a helpful AI assistant."
    },
    {
        "role": "user",
        "content": "What is quantization?"
    }
]

formatted = tokenizer.apply_chat_template(
    messages,
    tokenize=False,
    add_generation_prompt=True,
)

print(formatted)
```
